### Writing our first training loop 

**Training a neural network**
1. Create a model
2. Choose a loss function
3. Define a dataset
4. Set an optimizer 
5. Run a training loop:
    - Calculate a loss (Forward Pass)
    - Compute gradients (backpropogation)
    - Updating model parameters

The *Training Loop* repeats multiple times, it allows for greater flexibility and control, giving us the option to customizes different elements.
