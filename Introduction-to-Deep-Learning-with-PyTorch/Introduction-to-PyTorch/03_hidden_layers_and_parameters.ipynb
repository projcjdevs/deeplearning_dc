### Stacking layers with nn.Sequential()

A PyTorch container for stacking layers in sequence.

- This network takes input, passes it to each linear layer in sequence, and returns output.
- In this case, the layers within nn.Sequential() are hidden layers.

```python
# Creating a network with three linear layers
import torch
import torch.nn as nn

model = nn.Sequential(
    nn.Linear(n_features, 8),
    nn.Linear(8, 4), # Hidden layer
    nn.Linear(4, n_classes)
)
```

> ***n_features*** represents number of input features, while ***n_classes*** represents the number of output classes.

You can add as many hidden layers as you like, as long as the *input* dimension of each layer, matches the *output* dimension of the previous one.

```python
# Example

model = nn.Sequential(
    nn.Linear(10, 18), # Input layer
    nn.Linear(18, 20), # Hidden layer
    nn.Linear(20, 5)  # Output layer
)
```

### Layers are made of *Neurons*

Networks are considered ***fully connected*** when each neurons links to all neurons in the previous layer.

A neuron in a linear layer:
- Performs a linear operation using all neurons from the previous layer
- Has N+1 parameters: ***N*** from **inputs** and ***1*** for **bias**
---

### Parameters and Model Capacity

More hidden layers in a network means more parameters, and more parameters means **higher** ***model capacity*** -- Higher capacity models can handle more complex datasets but may take longer to train.

An effective way to assess a model's capacity is by calculating its total number of parameters.

```python
# Given the following model:

model = nn.Sequential(nn.Linear(8, 4),
                      nn.Linear(4, 2))
```

**Manual Parameter Calculation:**
- First layer has 4 neurons ((8,4) <- the 8 are input features that are connected to 4, which is considered as the actual **neurons**), and each neurons has 8+1 parameters (*Why?* - 8 input features OR weights + 1 bias, as explained in the previous notebook, hence 8+1). 9 parameters x 4 neurons = ***36 parameters***.
- Second layer has 2 neurons, each neuron has 4+1 parameters. 5 parameters x 2 neurons = ***10 parameters***
- Adding both layers together, this mdoel has ***46 learnable parameters***.

> You can calculate this with PyTorch, using the **.numel()** function, it returns the number of elements in the tensor.

```python
# Calculate the number of parameters in the model:

total = 0
for parameter in model.parameters():
    total += parameter.numel()
print(total)
```

**Balancing Complexity and Efficiency**

Understanding parameter count helps us balance model complexity and efficiency, too many parameters can lead to long training times or overfitting while too few might limit learning capacity.
