### Using derivatives to update model parameters
 
Let's now see how we can minimize ***loss***.

We know a model predicts poorly when loss is **high**, we can use ***derivatives*** or ***gradients*** to minimize this loss.

**An analogy for derivatives**
Imagine the loss function as a *valley*, the **derivative** represents the ***slope***, how steeply the curve rises or falls.
- **Steep slopes (Red Arrows)**:
    - Large steps, **high** derivatives
- **Gentler slopes (Green Arrows)**:
    - Small steps, **low** derivatives
- **Valley floor (Blue Arrow)**:
    - Flat, derivative is **zero** - This point is the loss function's minimum, which we aim to reach.

### **Convex and non-convex functions**

1. Convex functions has ***one*** global minimum

<img src="../images/convex.png" width="500">

2. Non-convex functions such as ***loss functions*** have multiple local minima, where the value is lower than nearby points, but not the lowest overall. When minimizing loss functions, we aim to locate the global minimum where x is approximately 1. 

<img src="../images/non-convex.png" width="500">



**Connecting derivatives and model training**
During training, we run a forward pass on the features and compute loss by comparing predictions to the target value.
> Recall that layer weights and biases are randomly initialized when a model is created, we update them during training using a ***backward pass or backpropogation***

In deep learning, **derivatives** are known as **gradients**, we compute the loss function gradients and use them to update the model parameters, including weights and biases with ***backpropogation***, repeating until the layers are tuned.

During backpropogation, we consider a network of three linear layers, we can calculate local loss gradients with respect to each layer's parameters.
- We first calculate loss gradients with respect to L2 parameters
- to L2 to compute L1 gradients
- repeat for all layers (L1, L0)

### Gradient Descent
We use a mechanism called gradient descent to find the global minimum of loss functions
- For non-convex functions, we will use this.
- PyTorch simplifies this with **optimizers**
    - Stochastic gradient descent (SGD)

```python
import torch.optim as optim

# Create the optimizer
optimizer = optim.SGD(model.parameters(), lr=0.001)

# Perform parameter updates
optimizer.step() # autumatically calculates gradients and updates model parameters
```
