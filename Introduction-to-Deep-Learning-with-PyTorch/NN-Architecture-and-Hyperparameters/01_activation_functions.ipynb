## Discovering Activation Functions

*Activation Functions* adds ***non-linearity*** to the network:
- ***Sigmoid*** - used for binary classifications.
- ***Softmax*** - used for multi-class classifications.

A network can learn more **complex** relationships with non-linearity.
- We will call the output of the **last** linear layer the ***pre-activation output*** which will pass the activation functions to obtain the transformed output.


> Note: Non-linearity is added to let network layers learn complex patterns, without it, a network with many layers would just act as a simple single-layer linear model -- EXACT same function from the attention is all you need research paper.

---

### **Sigmoid Function**
- Widely used for binary classification problems

Let's say we're classifying an animal if its a mammal or not >
- Input: 
    - Limbs: 4
    - Eggs: 0
    - Hair: 1
- Pre-activation output:
    - 6 (output of the linear layer - not yet interpretable) = *passed into the sigmoid activation function*.
    - we will obtain a value between 0 and 1.
- Activation:
    - if output is >0.5, class label = 1 (mammal)
    - if output is <=0.5, class label = 0 (not mammal)


```python
# Shows activated output immediately - has a 1D layer only
import torch
import torch.nn as nn

input_tensor = torch.tensor([[6]])
sigmoid = nn.Sigmoid()
output_tensor = sigmoid(input_tensor)
print(output_tensor)
```

```python
# Activation as the last layer
model = nn.Sequential(
    nn.Linear(6, 4), # First Linear Layer
    nn.Linear(4, 1), # Second Linear Layer
    nn.Sigmoid() # Sigmoid activation function
)

# Note: Sigmoid as a last step in the network of linear layers is equivalent to traditional logistic regression.
```

### **Softmax Function**
- Another popular activation function
- for ***multiclass*** classification involving more than two class labels

Let's say we have three classes:
- Bird (0)
- Mammal (1)
- Reptile (2)

**Softmax Function** takes three-dimensional as input and outputs the ***same*** shape

It outputs a probability distribution:
- Each element is a probability (it's bounded between 0 and 1).
- The sum of the output vector is **equal** to 1.


---

```python
import torch 
import torch.nn as nn

# Create an input tensor
input_tensor = torch.tensor(
    [[4.3, 6.1, 2.3]]
)

# Apply softmax along the last dimension
probabilities = nn.Softmax(dim=-1) 
output_tensor = probabilities(input_tensor)
print(output_tensor)
```

**Getting acquainted with softmax**
- **(dim = -1)** - Indicates **softmax** is applied to the input tensor's **last dimension**.
- **(nn.Softmax())** - Similar to **sigmoid function**, this function can be used as last step in **nn.Sequential()**.
