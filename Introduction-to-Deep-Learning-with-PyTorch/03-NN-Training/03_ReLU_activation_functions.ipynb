### ReLU Activation Functions

**Sigmoid and Softmax Functions**
- Some activation functions can shrink gradients too much, making training inefficient. So far, we've worked with two activation functions: *sigmoid* and *softmax*, which are typically used in a model's final layer.

> Sigmoid = **Binary** classification | Softmax = **Multi-Class* classification

---

**Limitations of the sigmoid function**
- Outputs are bounded between 0 and 1
- Usable anywhere in a network, however, the **Gradients:**
    - Are very small ofr large and small values of x
    - This phenomenon is called ***saturation***, during backpropogation, this becomes problematic because each gradients depends on the previous one, when the gradients are extremely small, they fail to update the weights effectively, leading to ***the vanishing gradients problems***, and it can make training deep networks very difficult.

**Limitations of the softmax function**
- also produces bounded outputs between 0 and 1, also suffers from **saturation** in a similar way, therefore, both of these activation functions are not ideal for hidden layers, and are best used in the **last** layers only.

We'll discover two widely used activation functions designed for use between linear layers or in hidden layers.

---

### ReLU

**Re**ctified **L**inear **U**nit (ReLU)

- f(x) = max(x, 0)
- For **positive** inputs: output equals input
- For **negative** inputs: output is 0

This function has no upperbound, and its gradients do not approach 0 for large values of x, which helps overcome the **vanishing gradients** problem


```python
import torch.nn as nn

# ReLU in PyTorch:

relu = nn.ReLU()
```

### Leaky ReLU

Is a variation of the **ReLU** function

- **Postive** inputs behave exactly like ReLU
- **Negative** - it multiplies the input by a small coefficient (default 0.01 in PyTorch) 

This ensures the gradient for negative inputs remain non-zero, preventing neurons from completely stopping learning, which can happen with standard ReLU. 


```python
# Leaky ReLU in PyTorch:

leaky_relu = nn.LeakyReLU(
    negative_slope = 0.05 # The negative_slope parameter controls the coefficient applied to negative inputs
)
```
