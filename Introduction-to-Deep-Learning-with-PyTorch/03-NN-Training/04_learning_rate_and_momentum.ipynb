### Learning Rate & Momentum

**Updating weights with SGD**
- Training a neural network means solving an optimization problem by minimizing the loss function and adjusting model parameters.

>**Stochastic Gradient Descent (SGD) Optimizer**
- Recall this as the optimizer we used to find the global minimum of loss functions. The optimizer takes the model's parameters along with two key arguments:
    1. Learning Rate: controls the step size of updates
    2. Momentum: adds inertia to help the optimizer moves smoothly and avoids getting stuck.

Understanding their impact helps us optimize efficiency.

Let's try to find the minimum of a U-shaped function. We start at *x=-2* and run the **SGD optimizer** for ten steps. After these steps, we observe that the optimizer is close to the minimum.
- We can also note that as we *approach* the *minimum*, the step size gradually ***decreases***
- This happens because the step size is the gradient multiplied by the learning rate. Since the function is less steep near zero, the gradient, and thus the step size, gets smaller.
